### Preprocessing Data

```python
# Separate features (X) and labels (y) for training data
X_train = df.drop(df.columns[0], axis=1)
y_train = df[df.columns[0]]

# Separate features (X) and labels (y) for test data
X_test = df_test.drop(df_test.columns[0], axis=1)
y_test = df_test[df_test.columns[0]]

```

```python
# Normalize pixel values to be between 0 and 1
X_train = X_train.astype("float32")/ 255.0
X_test = X_test.astype("float32") / 255.0

```

```python
X_train_img = X_train.values.reshape(-1, 28, 28)
X_test_img = X_test.values.reshape(-1, 28, 28)

print(f"Shape of X_train_img: {X_train_img.shape}")
print(f"Shape of X_test_img: {X_test_img.shape}")
```

```python
X_train_img
```

```python
y_train_cat= to_categorical(y_train, num_classes=10)
y_test_cat = to_categorical(y_test, num_classes=10)
```

```python
perceptron= Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10, activation='softmax')
])
```

```python
perceptron.compile(optimizer='sgd', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_percp = perceptron.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_cat),verbose=1)
```

```python
acc_percp= perceptron.evaluate(X_test_img,y_test_cat,verbose=0)[1]
acc_percp
```

```python
perceptron= Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),

    Dense(10, activation='softmax')
])
```

```python
perceptron.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_percp = perceptron.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_cat),verbose=1)
```

```python
acc_percp= perceptron.evaluate(X_test_img,y_test_cat,verbose=0)[1]
acc_percp
```

```python
X_train_cnn= X_train_img.reshape(-1,28,28,1)
X_test_cnn= X_test_img.reshape(-1,28,28,1)
```

```python
cnn = Sequential([
    Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),

    Conv2D(64, kernel_size=(3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dropout(0.5),

    Dense(10, activation='softmax')
])
```

```python
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_cn = cnn.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_cat),verbose=1)
```

```python
acc_cnn= cnn.evaluate(X_test_img,y_test_cat,verbose=0)[1]
acc_cnn
```

```python

def plot_training(history, title):
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(history.history['accuracy'], label="Train")
    plt.plot(history.history['val_accuracy'], label="Val")
    plt.title(f"{title} Accuracy")
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(history.history['loss'], label="Train")
    plt.plot(history.history['val_loss'], label="Val")
    plt.title(f"{title} Loss")
    plt.legend()
    plt.show()
```

```python
plot_training(history_percp, "Perceptron")
```

```python


plot_training(history_cn, "CNN")
```

```python
def show_side_by_side(models, model_names, X, X_cnn, y_true, n=5):
    idxs = np.random.choice(len(X), n, replace=False)
    plt.figure(figsize=(15, 6))
    for i, idx in enumerate(idxs):
        plt.subplot(2, n, i+1)
        plt.imshow(X[idx].reshape(28, 28), cmap="gray")
        plt.axis("off")
        plt.title(f"True: {y_true[idx]}")
        preds = [np.argmax(model.predict(X_cnn[idx].reshape(1, 28, 28, 1) if name == "CNN" else X[idx].reshape(1, 28, 28)))
                 for model, name in zip(models, model_names)]
        plt.subplot(2, n, n+i+1)
        plt.axis("off")
        plt.title("\n".join(f"{n}: {p}" for n, p in zip(model_names, preds)))
    plt.tight_layout()
    plt.show()
```

```python
show_side_by_side([perceptron, cnn], ["Perceptron", "ANN", "CNN"], X_test_img, X_test_cnn, y_test, 5)
```
