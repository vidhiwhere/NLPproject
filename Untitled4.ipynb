```python

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
```

```python
sentences = [
 "I love this product",
 "This movie made me smile",
 "Service was friendly and quick",
 "Today felt bright and happy",
 "This is the best day",
 "Absolutely fantastic experience",
 "I enjoyed every single moment",
 "Great job, well done",
 "The food tasted delicious",
 "Totally recommend to everyone",
 "Very satisfied with results",
 "This worked better than expected",
 "Amazing quality and value",
 "Such a pleasant surprise",
 "I feel positive about this",
 "I hate this product",
 "This movie bored me",
 "Service was rude and slow",
 "Today was cold and lonely",
 "This is the worst day",
 "Terrible experience overall",
 "I regret buying this",
 "Very disappointed with results",
 "The food tasted awful",
 "Do not recommend this",
 "It broke after one use",
 "Not worth the money",
 "Utterly frustrating and annoying",
 "I feel negative about this",
 "Such a waste of time",
]
labels = [1]*15 + [0]*15
labels = np.array(labels)
```

```python
vocab_size =2000
tok = Tokenizer(num_words= vocab_size, oov_token="<OOV>")
tok.fit_on_texts(sentences)
seqs=tok.texts_to_sequences(sentences)
maxlen= max(len(s) for s in seqs)
X= pad_sequences(seqs, maxlen=maxlen, padding="post")
y= labels
```

```python
maxlen
```

```python
X[0]
```

```python
embed_dim=16
rnn_units=8
```

```python
inp = Input(shape=(maxlen,),dtype="int32", name='input')
x= Embedding(input_dim=vocab_size, output_dim=embed_dim,mask_zero=True, name='embed')(inp)

rnn=SimpleRNN(units =rnn_units,return_sequences=True, return_state=True,name='simple_rnn')

# The RNN returns a tuple (output_sequence, last_state). We need to select one.
output_sequence, last_state = rnn(x)
out=Dense(1, activation='sigmoid', name='output')(last_state)

model= Model(inputs=inp , outputs= out)
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
model.summary()
```

```python
model.fit(X,y, epochs=25,batch_size=8,verbose=1)
```

### Visualizing RNN Hidden States

To visualize the 'pre time stamps' (which correspond to the hidden states at each time step, `output_sequence`) and the 'hidden states' (which usually refers to the final hidden state, `last_state`), we need to create a new model that outputs these intermediate tensors from our `simple_rnn` layer.

```python
vis_model = Model(inputs=inp, outputs=[output_sequence, last_state])
vis_model.summary()
```

Now, let's pick a couple of sample sentences, one positive and one negative, to see how their hidden states evolve.

```python
import matplotlib.pyplot as plt

# Choose a positive and a negative sentence
sample_sentences = [
    "I love this product", # Positive
    "I hate this product"  # Negative
]

# Tokenize and pad the sample sentences
sample_seqs = tok.texts_to_sequences(sample_sentences)
sample_X = pad_sequences(sample_seqs, maxlen=maxlen, padding="post")

# Get the hidden states from the visualization model
output_sequences, last_states = vis_model.predict(sample_X)

print("Shape of output_sequences (hidden states at each timestep):", output_sequences.shape)
print("Shape of last_states (final hidden states):", last_states.shape)
```

We have `output_sequences` which contains the hidden states for each word (timestep) in the sentence, and `last_states` which is the final hidden state after processing the entire sentence. Let's visualize the `output_sequences` for our sample sentences.

```python
# Plotting the hidden states at each timestep for the first sample sentence
plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
plt.plot(output_sequences[0])
plt.title(f'Hidden States over Timesteps for: "{sample_sentences[0]}"')
plt.xlabel('Timestep')
plt.ylabel('Hidden State Value')
plt.legend([f'RNN Unit {i+1}' for i in range(rnn_units)], loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(output_sequences[1])
plt.title(f'Hidden States over Timesteps for: "{sample_sentences[1]}"')
plt.xlabel('Timestep')
plt.ylabel('Hidden State Value')
plt.legend([f'RNN Unit {i+1}' for i in range(rnn_units)], loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid(True)

plt.tight_layout()
plt.show()
```

These plots show how the values of each RNN unit (each line represents one unit) change as the network processes each word (timestep) in the sentence. You can observe how the pattern differs between the positive and negative sentences, especially towards the end.

Let's also visualize the final hidden states (`last_states`) as a bar plot to compare the summary representations for the positive and negative sentences.

```python
plt.figure(figsize=(10, 5))
bar_width = 0.35
index = np.arange(rnn_units)

plt.bar(index, last_states[0], bar_width, label=sample_sentences[0], alpha=0.8)
plt.bar(index + bar_width, last_states[1], bar_width, label=sample_sentences[1], alpha=0.8)

plt.xlabel('RNN Unit')
plt.ylabel('Final Hidden State Value')
plt.title('Comparison of Final Hidden States for Sample Sentences')
plt.xticks(index + bar_width / 2, [f'Unit {i+1}' for i in range(rnn_units)])
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

This bar plot shows the final state of each RNN unit after processing the entire sentence. Differences in these final states are what the subsequent Dense layer uses to classify the sentiment.
